```python
# Session 4 — Mini RAG Demo
# Completed by Rishikesh Munnaluri
import re

documents = [
    "CS 202 Machine Learning. Midterm Exam: March 15th, 2026. Weight: 30%.",
    "CS 202 Machine Learning. Final project submission: April 28th, 2026. Weight: 40%.",
    "CS 202 Machine Learning. Weekly quizzes contribute 10% of the course grade."
]

def tokenize(text):
    return set(re.findall(r"[a-z0-9]+", text.lower()))

def retrieve(query, docs):
    q = tokenize(query)
    scored = []
    for doc in docs:
        score = len(q & tokenize(doc))
        scored.append((score, doc))
    return max(scored, key=lambda x: x[0])[1]

def rag_answer(query):
    context = retrieve(query, documents)
    date_match = re.search(r"(March|April|May|June|July|August|September|October|November|December)\s+\d+(?:st|nd|rd|th)?,\s+\d{4}", context)
    answer = date_match.group(0) if date_match else context
    return context, answer

query = "When is the ML midterm?"
context, answer = rag_answer(query)
print("Retrieved context:", context)
print("Answer:", answer)
```

### Observation
RAG retrieves relevant external context at inference time. The model does not need to be retrained on the document.
