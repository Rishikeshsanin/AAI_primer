```python
# Session 2 — Function / Tool Calling
# Completed by Rishikesh Munnaluri
import json

def get_weather(city: str):
    weather = {
        "mumbai": "32C, Humid",
        "delhi": "28C, Smoggy",
        "bangalore": "27C, Sultry",
        "chennai": "33C, Humid"
    }
    return weather.get(city.lower(), "Data not available")

tool_schema = [{
    "type": "function",
    "function": {
        "name": "get_weather",
        "description": "Get current mock weather of a city",
        "parameters": {
            "type": "object",
            "properties": {"city": {"type": "string"}},
            "required": ["city"]
        }
    }
}]

def simulate_tool_call(user_query):
    known = ["mumbai", "delhi", "bangalore", "chennai"]
    city = next((c for c in known if c in user_query.lower()), None)
    if not city:
        return {"answer": "Please specify a supported city."}
    tool_call = {"name": "get_weather", "arguments": {"city": city.title()}}
    result = get_weather(city)
    return {"tool_call": tool_call, "tool_result": result,
            "final_answer": f"The weather in {city.title()} is {result}."}

result = simulate_tool_call("What's the weather like in Bangalore?")
print(json.dumps(result, indent=2))
```

### Observation
The LLM does not directly execute Python. It produces a structured tool-call request; the host program executes the function and returns the result.
