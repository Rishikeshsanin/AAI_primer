```python
# Session 1 — From Prediction to Action
# Completed by Rishikesh Munnaluri

def predict_spam(email_text):
    spam_terms = ["won", "prize", "free money", "$1,000,000"]
    text = email_text.lower()
    score = sum(term in text for term in spam_terms)
    confidence = min(0.70 + 0.10 * score, 0.99)
    return {"label": "SPAM" if score else "NOT_SPAM", "confidence": round(confidence, 2)}

class MockLLM:
    def ask(self, instruction):
        return {
            "reply": "Thanks for reaching out. I am currently busy and will respond later.",
            "action": "MOVE_TO_TRASH"
        }

llm = MockLLM()

def agentic_handler(email_text):
    prediction = predict_spam(email_text)
    if prediction["label"] == "SPAM" and prediction["confidence"] > 0.9:
        return {
            "prediction": prediction,
            "agent_action": llm.ask("Draft a polite busy reply and move to trash.")
        }
    return {"prediction": prediction, "agent_action": "KEEP_IN_INBOX"}

sample = "Congratulations! You won a $1,000,000 prize and free money."
print(agentic_handler(sample))
```

### Observation
A conventional ML function only predicts a label. An agentic workflow uses the prediction to decide and perform an action.
