```python
# Session 3 — ReAct Loop (Reason + Act)
# Completed by Rishikesh Munnaluri

def get_weather(city):
    return {
        "mumbai": {"temp": 32, "condition": "Humid"},
        "delhi": {"temp": 28, "condition": "Smoggy"},
        "bangalore": {"temp": 27, "condition": "Sultry"}
    }.get(city.lower())

def react_compare_weather(city_a, city_b):
    trace = []
    trace.append(f"Thought: I need weather for {city_a} and {city_b}.")
    trace.append(f"Action: get_weather('{city_a}')")
    a = get_weather(city_a)
    trace.append(f"Observation: {a}")
    trace.append(f"Action: get_weather('{city_b}')")
    b = get_weather(city_b)
    trace.append(f"Observation: {b}")
    warmer = city_a if a["temp"] > b["temp"] else city_b
    trace.append(f"Final Answer: {warmer.title()} is warmer ({max(a['temp'], b['temp'])}C).")
    return trace

for line in react_compare_weather("mumbai", "delhi"):
    print(line)
```

### Observation
ReAct enables an agent to perform multiple tool calls before producing the final response. This is more capable than a single one-shot function call.
